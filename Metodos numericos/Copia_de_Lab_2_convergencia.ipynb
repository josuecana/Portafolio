## Laboratory 2: Convergence

* In 1593, François Viète published what is now known as Viètes_formula, an infinite product (rather than an infinite sum, which is more typically used in π calculations): https://en.wikipedia.org/wiki/Pi

$$\frac{2}{\pi} = \frac{\sqrt{2}}{2}\times \frac{\sqrt{2+\sqrt{2}}}{2}\times \frac{\sqrt{2+\sqrt{2+\sqrt{2}}}}{2}\times \cdots$$

* The Leibniz series is a simple way to approximate the value of $\pi$:

$$
\pi \approx 4 \sum_{n=0}^{N} \frac{(-1)^n}{2n + 1}
$$



Some infinite series for π converge faster than others. Given the choice of two infinite series for π, mathematicians will generally use the one that converges more rapidly because faster convergence reduces the amount of computation needed to calculate π to any given accuracy.


**Exercise 1**:

Make a figure where is compared the values obtained for the two methods for calculating $\pi$ as a function of the number of performed iterations.

Which method reaches a faster convergence? Which approximation takes less terms for achieving a good accuracy of the pi number?


```python
import numpy as np
import math as math
leibniz = []
Viètes_formula = []
N = 100
suma = 0
for i in range(N + 1):
    suma += ((-1)**i)/(2*i+1)
    leibniz.append(4*suma)
raiz = np.sqrt(2)
multi = raiz/2
for i in range(N+1):
  raiz = np.sqrt(2+raiz)
  multi *= raiz/2
  Viètes_formula.append(2/multi)
```

```python
import matplotlib.pyplot as plt
X = np.linspace(1,N,N+1)
pi = np.linspace(np.pi,np.pi,N+1)
plt.plot(X,Viètes_formula, label ="Viètes_formula", color = "red")
plt.plot(X,pi, label ="π" ,color = "black")
plt.plot(X,leibniz, label ="formula de leibniz", color = "green")
plt.ylim(2.96,3.32)
plt.legend()
plt.show()
```

```python
error = []
errorl = []
for i in range(len(Viètes_formula)):
  error_ad = np.abs(Viètes_formula[i]-np.pi)
  error_le = np.abs(leibniz[i]-np.pi)
  error.append(error_ad)
  errorl.append(error_le)

plt.title("Error absoluto")
plt.semilogy(X,error, label ="|E_ex-E_t| (Viètes_formula)",color = "red")
plt.plot(X,errorl, label ="|E_ex-E_t| (leibniz)", color = "green")
plt.legend()
plt.show()
```

Podemos concluir que la formula de vities es converge mucha mas rapido y tiene un menor error que la de leibniz
