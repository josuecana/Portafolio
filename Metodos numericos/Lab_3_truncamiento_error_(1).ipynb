* Escoja una función, hacerle la expansión de Taylor alrededor de un punto $x_0$ hasta un orden N.

* Dibujar la expasión para varios ordenes $N$ en conjunto con la función según numpy.
* ¿Qué puede decir respecto al error de truncamiento?


El truncamiento muestra que en los limites de tangente (las axintotas) el error es muy grande pero en el resto de la funcion la serie de taylor muestra ser bastante precisa,

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
fig, ax = plt.subplots(1,2,figsize=(15,5))

x = sp.Symbol('x')
x0 = 0

h = np.linspace(-np.pi/2,np.pi/2,200)
for n in range(2,20,2):
  t = sp.tan(x).series(x,x0,n).removeO()
  t = sp.lambdify(x,t)
  yh = np.tan(h)
  yt = t(h)
  lista = np.abs(yh - yt)
  ax[0].plot(h,yt,label =f"n = {n}")
  ax[1].plot(h,lista,label=f"n = {n}")
ax[0].set_title("Serie de tailor vs valor teorico")
ax[0].plot(h,yh,label = "valor teorico" ,color = "green")
ax[0].set_ylim(-10,10)
ax[0].legend()
ax[1].set_title("Error")
ax[1].set_yscale("log")
ax[1].legend()
plt.show()
```

Podemos ver como ha medida que avanza la serie esta se acerca mas al valor toerico y reducciondo su error de truncamiento, el error muestra una presicion muy alta desde n=4, pero ha medida que se aleja del origen este error tiene ciertos aumentos, esto puedo deberse a las asintotas que tiene la tangente ya que en estos lugares toma valores muy elevados la funcion, entonces esto puedo explicar ese aumento en el error.
