```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def exp_d(t, landa=0.01):
    return landa * np.exp(-landa * t)

lorenziana = lambda x: (1/(np.pi*100)) *(100**2)/(100**2+(x-100)**2)

def muestreo(tamaño, repeticiones):
    muestras = np.random.exponential(1/0.01, size=(repeticiones, tamaño))
    return list(np.mean(muestras, axis=1))

caso_1 = muestreo(10000, 10)
caso_2 = muestreo(100000, 50)
caso_3 = muestreo(1000000, 100)

def muestreo2(tamaño, repeticiones):
    muestras = np.random.standard_cauchy(size=(repeticiones,tamaño))*100+100
    return list(np.mean((muestras), axis=1))

casol_1 = muestreo2(10000, 10)
casol_2 = muestreo2(100000, 50)
casol_3 = muestreo2(1000000, 100)
```

```python
bins = np.linspace(0, 500, 51)
x = np.linspace(100 - 4*100, 100 + 4*100, 100)

y = (1/(100*np.sqrt(2*np.pi))) * np.exp(-0.5*((x - 100)/100)**2)
fig, ax = plt.subplots(1, 3, figsize=(15, 4))

ax[0].hist(caso_1, bins=bins, density=True)
ax[0].plot(x,y,color="red")
ax[0].set_title('10k muestras, 10 reps')

ax[1].hist(caso_2, bins=bins, density=True)
ax[1].plot(x,y,color="red")
ax[1].set_title('100k muestras, 50 reps')

ax[2].hist(caso_3, bins=bins, density=True)
ax[2].plot(x,y,color="red")
ax[2].set_title('1M muestras, 100 reps')

plt.tight_layout()
plt.show()
```
