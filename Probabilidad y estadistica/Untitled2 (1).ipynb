```python
import numpy as np
import matplotlib.pyplot as plt
def dados(tiros):
  dados1 = [1,2,3,4,5,6]
  dados2 = [1,2,3,4,5,6]
  tiro = []
  for i in range(tiros):
    tiro.append(np.random.choice(dados1) + np.random.choice(dados2))
  return (tiro)
```

```python
i, j = np.meshgrid(np.arange(1,7), np.arange(1,7))
sumas = (i + j).flatten()

plt.hist(sumas, bins=np.arange(2,14)-0.5,label = "probilidad", density=True)
intento_1 = (dados(10000))
intento_2 = dados(1000000)
plt.hist(sumas, bins=np.arange(2,14)-0.5, density=True)
plt.hist(intento_1,bins=11,label = "tiros: 10000",density=True,alpha = 1,color='teal', edgecolor='black', rwidth=0.85)
plt.hist(intento_2,bins=11,label = "tiros: 1000000",density=True,alpha = 0.5,color='red', edgecolor='black', rwidth=0.85)
plt.title("Dados normales")
plt.legend()
plt.show()
```

```python
def dados_cuan(tiros):
  espacio = []
  suma = []
  for i in range(1,7):
    for n in range(i,7):
      if i!=n:
        espacio.append([i,n])
  for l in range(tiros):
      k = np.random.randint(len(espacio))
      suma.append(espacio[k][0]+espacio[k][1])
  return suma
```

```python
intento_1 = dados_cuan(10000)
intento_2 = dados_cuan(1000000)
i, j = np.meshgrid(np.arange(1,7), np.arange(1,7))
mask = i < j
sumas = (i + j)[mask]

plt.hist(sumas, bins=np.arange(2,14)-0.5,label = "probabilidad", density=True)
plt.hist(intento_1,bins=9,label = "tiros: 10000",density=True,alpha = 1,color='teal', edgecolor='black', width=0.6)
plt.hist(intento_2,bins=9,label = "tiros: 1000000",density=True,alpha = 0.5,color='red', edgecolor='black', width=0.6)
plt.legend()
plt.title("Dados cuanticos")
plt.show()
```

```python
def dados_fer(tiros):
  espacio = []
  suma = []
  for i in range(1,7):
    for n in range(i,7):
        espacio.append([i,n])
  for l in range(tiros):
      k = np.random.randint(len(espacio))
      suma.append(espacio[k][0]+espacio[k][1])
  return suma
intento_1 = dados_fer(10000)
intento_2 = dados_fer(1000000)
i, j = np.meshgrid(np.arange(1,7), np.arange(1,7))
mask = i <= j

sumas = (i + j)[mask]

plt.hist(sumas, bins=np.arange(2,14)-0.5,label="Probabilidad", density=True)
plt.hist(intento_1,bins=11,label = "tiros: 10000",density=True,alpha = 1,color='teal', edgecolor='black', rwidth=0.85)
plt.hist(intento_2,bins=11,label = "tiros: 1000000",density=True,alpha = 0.5,color='red', edgecolor='black', rwidth=0.85)
plt.legend()
plt.title("Dados bosonicos")
plt.show()
```
